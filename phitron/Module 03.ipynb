## list comprehension

```python
even = []

for i in range(100):
    if i %2 ==0:
        even.append(i)
print(even)
```

```python
even.clear()
even
```

```python
num=[x for x in range(10)]
num
```

```python
num2=[x*2 for x in range(10)]
num2
```

```python
even2=[x for x in range(10) if x%2==0]
even2
```

```python
odd = [x for x in range(100) if x%2!=0]
print(odd)
```

## list comprehension in string

```python
fruits = ["apple", "orange", "lichi"]

capitaize_fruits =[]

for fruit in fruits:
    capitaize_fruits.append(fruit.capitalize())

capitaize_fruits
```

```python
upper_fruit = [fruit.upper() for fruit in fruits]
upper_fruit
```
