# For Loop

```python
for i in range(1,11):
    print("hello world", i)
```

# While loop


```python
count=1

while count < 10:
    print("hello world", count)
    count+=1
```

## Break & continue

```python
sum=0

for i in range(1, 11):
    if i%2 ==0:
        print(i, "is skiping")
        continue
    sum +=i
    print(i, "is added to the sum")
print(sum)
```

```python
accuracy=95

for i in range(20):
    accuracy+=1
    print(accuracy)
    if accuracy == 100:
        break
```

```python
accuracy = 78
while True:
    if accuracy == 100:
        break
    accuracy +=1
```
