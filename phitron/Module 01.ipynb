# Operators

```python
x=10 ; y=3

sum = x+y
sub = x-y
mult = x*y
div = round(x/y, 2)

rem = x%y

print(sum, sub, mult, div, rem)
```

```python
floor_div = x//y
power = x**y

print(floor_div, power)
```

```python
x=10 ; y=3


greater_than = x>y
less_than = x<y
equal = x==y


print(greater_than, less_than, equal)
```

```python
greater_than_equal = x>=y
less_than_equal = x<=y
not_equal = x!=y

print(greater_than_equal, less_than_equal, not_equal)
```
