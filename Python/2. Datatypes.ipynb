Python supports 3 categories data

* Basic (int, float, complex, boolean, string)
* Container type(list, tuples, sets, dictionary)
* user-defined types

```python
# integer
print(4)
```

```python
print(1e308) #max int
```

```python
print(1e309)
```

```python
# boolean
print(True)
print(False)
```

```python
# complex
print(4+5j)
```

```python
# strings

print("Dhaka")
```

```python
# list
print([1, 2, 3, 4, 5])
```

```python
# tuple
print((1, 2, 3, 4, 5))
```

```python
# sets
print({1, 2, 3, 4, 6})
```

```python
# dict
print({"name": "Ashis", "age": 25, "gender": "male"})
```
