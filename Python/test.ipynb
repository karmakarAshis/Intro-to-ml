```python
# Question 5
number = int(input())

original_number = number
reversed_number = 0

while number > 0:
    digit = number % 10
    reversed_number = reversed_number * 10 + digit
    number = number // 10

if original_number == reversed_number:
    print("Miror Number")
else:
    print("Not Mirror Number")

```

```python
# q3

with open("./sample_data.txt", "r") as file:
    strings = file.read()
    strings = strings.lower().split()

word_counts = {}    

for word in strings:
    word_counts[word] = word_counts.get(word, 0) + 1

max_word = None
max_count = 0

for word in word_counts:
    if word_counts[word] > max_count:
        max_count = word_counts[word]
        max_word = word

print(max_word)
```
