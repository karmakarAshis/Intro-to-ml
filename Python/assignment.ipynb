```python
inp = input()
numbers = inp.split()

numbs = [int(x) for x in numbers]
nr = []

for i in range(len(numbs)):
    if numbs[i] %3 ==0 and numbs[i] %5 !=0:
        nr.append(numbs[i])

print(nr)
```

```python
def compute(a, b):
    try:
        result = ((a**2)+(b**2))/(a-b)
    except Exception as Error:
        print(f"{Error}: Division not possible")
    else:
        return result



inp = input()
numbers = inp.split()

a = int(numbers[0])
b = int(numbers[1])

compute(a, b)
```

```python
# ?q3
from functools import reduce

with open("./data.txt", "r") as file:
    
    strings = file.readlines()
    print(strings)
    # print(len(strings))

    number_of_words = list(map(lambda x: len(x.split()), strings))
    print(number_of_words)

    total_number_of_words = reduce(lambda x,y: x+y, number_of_words)
    print(total_number_of_words)

    strings = list(map(str.strip, strings))
    print(strings)
```
