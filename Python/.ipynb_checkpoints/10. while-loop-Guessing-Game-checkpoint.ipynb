```python
import random
```

```python
random.randint(1, 100)
```

```python
jackpot = random.randint(1, 100)

guess = int(input("Guess the number: "))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print("guess higher")
    else:
        print("guess lower")

    guess = int(input("Guess the number: "))
    counter += 1
print("Right Answer") 
print("took", counter, "attempt")
```
